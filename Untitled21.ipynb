```python
#Total Missing values
df.isna().sum()
```

```python
# Fill the missing values with the use of mean, median, mode
#Fill misssing values with the use of median in closed date
df['Closed Date'] = pd.to_datetime(df['Closed Date'], errors='coerce')
df['Closed Date'] = df['Closed Date'].fillna(df['Closed Date'].median())

#Fill the missing value with the use of mode in location type
df['Location Type'] = df['Location Type'].fillna(df['Location Type'].mode()[0])

#Fill the missing values with the use of mode in Incident Zip
df['Incident Zip'] = df['Incident Zip'].fillna(df['Incident Zip'].mode()[0])

#Fill the missing values with the use of mode in incident address
df['Incident Address'] = df['Incident Address'].fillna(df['Incident Address'].mode()[0])

#Fill the missing values with the use of mode in street name
df['Street Name'] = df['Street Name'].fillna(df['Street Name'].mode()[0])

#Fill the missing values with the use of mode in address type
df['Address Type'] = df['Address Type'].fillna(df['Address Type'].mode()[0])

#Fill the missing values with the use of mode in city
df['City'] = df['City'].fillna(df['City'].mode()[0])

#Fill the missing values with the use of mode in Landamark
df['Landmark'] = df['Landmark'].fillna(df['Landmark'].mode()[0])

#Fill the missing values with the use of mode in Facility Type
df['Facility Type'] = df['Facility Type'].fillna(df['Facility Type'].mode()[0])

#Fill the missing values with the use of mean in Due Date
df['Due Date'] = pd.to_datetime(df['Due Date'], errors='coerce')
df['Due Date'] = df['Due Date'].fillna(df['Due Date'].mean())

#Fill the missing values with the use of mode in Resolution Description
df['Resolution Description'] = df['Resolution Description'].fillna(df['Resolution Description'].mode()[0])

#Fill the missing values with the use of mean in BBL
df['BBL'] = df['BBL'].fillna(df['BBL'].mean())

#Fill the missing values with the use of median in Latitude
df['Latitude'] = df['Latitude'].fillna(df['Latitude'].median())

#Fill the missing values with the use of mean in Longitude
df['Longitude'] = df['Longitude'].fillna(df['Longitude'].mean())

#Fill the missing values with the use of mode in Location
df['Location'] = df['Location'].fillna(df['Location'].mode()[0])
```

```python
# Review Cleaned Dataset
df.isna().sum()
```

Univariate analysis analyse one variable at a one time.

```python
# Univariate Analysis
# Number of complaints by Borough
df['Borough'].value_counts().plot(kind='bar')
```

```python
# Univariate analysis of Top 10 complaints by complaint type
top10 = df['Complaint Type'].value_counts().head(10)
top10.plot(kind='bar')

plt.title('Top 10 Complaints by Complaint type')
plt.xlabel('Complaint Type')
plt.ylabel('Number of Complaints')
plt.xticks(rotation = 60)
plt.show()
```

**Bivariate** analysis analyse two variables itself.

```python
# Bivariate Analysis of Borough VS Complaint Type
# Select Top 5 complaint types
Top5 = df['Complaint Type'].value_counts().head(5).index

# Filter data
filtered_df = df[df['Complaint Type'].isin(Top5)]

# Create table of Borough VS Complaint Type
comparison = pd.crosstab(filtered_df['Borough'], filtered_df['Complaint Type'])

#plot
comparison.plot(kind='bar', figsize=(12,6))

plt.title('Top 5 Complaint Type by Borough')
plt.xlabel('Borough')
plt.ylabel('Number of Complaints')
plt.xticks(rotation = 45)
plt.show()
```

```python
# Correlation matrix for numeric columns
corr_matrix = df.select_dtypes(include=['int64','float64']).corr()
corr_matrix
```

```python
# Correlation Matrix of Heatmap
plt.figure(figsize=(10,6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')

plt.title('Correlation Matrix Heatmap')
plt.show()
```
